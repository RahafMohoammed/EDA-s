```python
sns.boxplot(x="ocean_proximity", y="median_house_value", data=df)
plt.xticks(rotation=45)
plt.show()
```

## متى أستخدم Box Plot؟

**Box Plot هو الجواب لما تسأل:**
"كيف يتوزع رقم معين عبر مجموعات مختلفة؟"

### القاعدة البسيطة:
عمود رقمي + عمود فئات = Box Plot

### أمثلة على الأسئلة:
- هل المجموعة X أغلى من Y؟
- وين التفاوت الأكبر بين المجموعات؟
- هل في قيم شاذة داخل كل مجموعة؟

### متى لا أستخدمه؟
- عمود رقمي بدون فئات ← Histogram
- علاقة بين رقمين ← Scatter Plot

```python
df['ocean_proximity'].value_counts().plot(kind='bar')
plt.title('Distribution of Ocean Proximity')
```

```python
df.corr(numeric_only=True)["median_house_value"].sort_values(ascending=False)
```

```python
df["rooms_per_house"] = df["total_rooms"] / df["households"]
df["bedrooms_per_room"] = df["total_bedrooms"] / df["total_rooms"]
df["people_per_house"] = df["population"] / df["households"]
```

```python
df.groupby("ocean_proximity")[["rooms_per_house", "bedrooms_per_room", "people_per_house"]].mean().round(2)
```

```python
df.groupby("ocean_proximity").mean(numeric_only=True).round(2)
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```
